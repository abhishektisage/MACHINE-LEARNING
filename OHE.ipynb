```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Female', 'Male'],
    'City': ['Delhi', 'Mumbai', 'Pune', 'Delhi'],
    'Age': [25, 30, 22, 35]
})
```

```python
encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(df[['Gender', 'City']])

```

```python
encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(['Gender', 'City'])
)
```

```python
final_df = pd.concat([df[['Age']], encoded_df], axis=1)

print(final_df)
```
